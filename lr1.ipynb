```python
def divide_number_digits(number: int) -> list[int]:
    array_digits: list[int] = []
    
    while number > 0:
        array_digits.insert(0, number % 10)
        number //= 10

    return array_digits

def combine_digits_number(array_digits: list[int]) -> int:
    number = 0
    multiplier = 1

    for digit in array_digits[::-1]:
        number += digit * multiplier
        multiplier *= 10

    return number
```

### Variant 28

```python
def to_four_digit_system(number: int) -> int:
    integer_part = number
    number_quaternary_system: int = 0
    multiplier = 1

    while integer_part > 0:
        if integer_part < 4:
            number_quaternary_system += integer_part * multiplier
            break

        digit = integer_part % 4
        number_quaternary_system += digit * multiplier
        integer_part //= 4
        multiplier *= 10

    return number_quaternary_system
```

```python
left_bound: int = 128; right_bound: int = 255
initial_number_N: int = random.randint(left_bound, right_bound)
initial_number_M: int = random.randint(left_bound, right_bound)
# initial_number_N = 224; initial_number_M = 173
initial_number_N_quaternary_system: int = to_four_digit_system(initial_number_N)
initial_number_M_quaternary_system: int = to_four_digit_system(initial_number_M)
initial_number_N_quaternary_system = 2132
initial_number_M_quaternary_system = 3010
answer: int = 0
array_digits: list[int] = []
carry = 0
multiplier = 1
```

```python
print(
    "Initial number N:", initial_number_N, 
    "\nInitial number M:", initial_number_M, 
    "\nInitial number N in quantenary system:", initial_number_N_quaternary_system,
    "\nInitial number M in quantenary system:", initial_number_M_quaternary_system
)

while initial_number_N_quaternary_system > 0:
    digit_N = initial_number_N_quaternary_system % 10
    digit_M = initial_number_M_quaternary_system % 10
    sum_digit_N_digit_M = digit_N + digit_M

    if carry == 1:
        sum_digit_N_digit_M += carry
        carry = 0

    if sum_digit_N_digit_M == 4:
        array_digits.append(0)
        carry = 1
    elif sum_digit_N_digit_M > 4:
        array_digits.append(sum_digit_N_digit_M % 4)
        carry = 1
    else:
        array_digits.append(sum_digit_N_digit_M)

    initial_number_N_quaternary_system //= 10
    initial_number_M_quaternary_system //= 10

if carry == 1:
    array_digits.append(carry)

for digit in array_digits:
    answer += digit  * multiplier
    multiplier *= 10

print("Answer:", answer)
```
