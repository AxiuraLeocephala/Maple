```python
import random 
from datetime import datetime
import math
from decimal import Decimal
```

```python
NUMBER_DAYS_WEEK= 7
DAYS_OF_WEEK = {
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday",
    7: "Sunday"
}
NUMBER_29_FEBREARY = 60 # 29 Febreary
NUMBER_MINUTES_IN_DAY = 1440
NUMBER_MINUTES_IN_HOUR = 60
NUMBER_SECONDS_IN_DAY = 86400
NUMBER_SECONDS_IN_HOUR = 3600
NUMBER_SECONDS_IN_MINUTES = 60
```

```python
def divide_number_digits(number: int) -> list[int]:
    array_digits = []
    
    while number > 0:
        array_digits.insert(0, number % 10)
        number //= 10

    return array_digits

def combine_digits_number(array_digits: list[int]) -> int:
    number = 0
    multiplier = 1

    for digit in array_digits[::-1]:
        number += digit * multiplier
        multiplier *= 10

    return number

# def length_number(number: int) -> int:
#     return len(divide_number_digits(number))

def decimal_number_to_another_base(decimal_number: int, base: int) -> int:
    number_another_base = 0
    multiplier = 1

    while decimal_number > 0:
        if decimal_number < base:
            number_another_base += decimal_number * multiplier
            break

        digit = decimal_number % base
        number_another_base += digit * multiplier
        decimal_number //= base
        multiplier *= 10

    return number_another_base

def another_base_number_to_decimal(number: int, base: int) -> int:
    digits_number_another_base = divide_number_digits(number)
    digits_number_another_base.reverse()

    return sum([d * (base ** i) for d, i in zip(
        digits_number_another_base, range(len(digits_number_another_base))
    )])

def bitwise_addition_numbers(a: int, b: int, base: int) -> int:
    carry = 0
    array_digits = []

    while a > 0:
        digit_N, digit_M = a % 10, b % 10
        sum_digit_N_digit_M = digit_N + digit_M

        if carry == 1:
            sum_digit_N_digit_M += carry
            carry = 0

        if sum_digit_N_digit_M == base:
            array_digits.append(0)
            carry = 1
        elif sum_digit_N_digit_M > base:
            array_digits.append(sum_digit_N_digit_M % base)
            carry = 1
        else:
            array_digits.append(sum_digit_N_digit_M)

        a //= 10; b //= 10
    
    if carry == 1: array_digits.append(carry)

    array_digits.reverse()

    return combine_digits_number(array_digits)
```
